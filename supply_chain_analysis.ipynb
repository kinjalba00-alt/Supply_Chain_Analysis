```python
import pandas as pd
df = pd.read_csv('quick_commerce_data_raw.csv')
```

```python
#null values
df.isnull().sum()
```

```python
df['Items_Count'] = df.groupby('Product_Category')['Items_Count'].transform(
    lambda x: x.fillna(x.median())
)
```

```python
df['Customer_Rating'] = df.groupby('Product_Category')['Customer_Rating'].transform(
    lambda x: x.fillna(x.median())
)
```

```python
df['Delivery_Partner_Rating'] = df.groupby('Product_Category')['Delivery_Partner_Rating'].transform(
    lambda x: x.fillna(x.median())
)
```

```python
df.isnull().sum()
```

```python
df['City'] = df.groupby('Product_Category')['City'].transform(
    lambda x: x.fillna(x.mode()[0] if not x.mode().empty else 'Unknown')
)
```

```python
df.isnull().sum()
```

```python
#removing spaces and convert snack case
df.columns = df.columns.str.lower()
df.columns = df.columns.str.replace(' ','_')
```

```python
df.columns
```

```python
#distance_km convert to a 4 labels
labels = ['Very_Near', 'Near', 'Far', 'Very_Far']
df['distance_group'] = pd.qcut(df['distance_km'], q=4, labels=labels)
df[['distance_km','distance_group']].head(10)
```

```python

df['partner_rating_bucket'] = pd.cut(
    df['delivery_partner_rating'],
    bins=[0, 2, 3.5, 5],
    labels=['Poor', 'Average', 'Good']
)
df[['delivery_partner_rating','partner_rating_bucket']].head(10)
```

```python
city_order_count = df['city'].value_counts()
median_count = city_order_count.median()

city_tier_map = city_order_count.apply(lambda x: 'High_Volume' if x > median_count else 'Low_Volume')
df['city_volume_tier'] = df['city'].map(city_tier_map)

df[['city','city_volume_tier']].head(10)
```

```python
import pandas as pd
from sqlalchemy import create_engine
from urllib.parse import quote_plus
import io, csv, time

# Step 0: Load data
df = pd.read_csv("quick_commerce_data_raw.csv")
print(df.shape)

# Step 1: Connection
username = "postgres"
password = "@Vaghela20"
host = "localhost"
port = 5432
database = "Delivery"

encoded_password = quote_plus(password)
encoded_username = quote_plus(username)

engine = create_engine(
    f"postgresql+psycopg2://{encoded_username}:{encoded_password}@{host}:{port}/{database}"
)

# Step 2: Fast COPY-based insert
def psql_insert_copy(table, conn, keys, data_iter):
    dbapi_conn = conn.connection
    with dbapi_conn.cursor() as cur:
        s_buf = io.StringIO()
        writer = csv.writer(s_buf)
        writer.writerows(data_iter)
        s_buf.seek(0)
        columns = ', '.join(f'"{k}"' for k in keys)
        sql = f'COPY "{table.name}" ({columns}) FROM STDIN WITH CSV'
        cur.copy_expert(sql=sql, file=s_buf)

start = time.time()
with engine.begin() as conn:
    df.to_sql("customer", conn, if_exists="replace", index=False, method=psql_insert_copy)

print(f"Done in {time.time() - start:.2f} seconds")
print("Data successfully loaded into table 'customer' in database 'Delivery'.")
```
